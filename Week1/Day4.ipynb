```python
import tiktoken
```

```python
encoding = tiktoken.encoding_for_model("gpt-4.1-mini")

tokens = encoding.encode("Hi my name is Jay and I like playing cricket.")

tokens
```

```python
for token in tokens:
    print(f"{token} = {encoding.decode([token])}")
```

```python
print(encoding.decode([1]))
print(encoding.decode([2]))
print(encoding.decode([3]))
```

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
OLLAMA_BASE_URL = "http://127.0.0.1:11434/v1"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')

response = ollama.chat.completions.create(
    model="gemma4:latest",
    messages=[
        {"role": "user", "content": "Hi, I am Jay. How are you?"}, 
        {"role" : "system", "content": "You are a helpful assistant that summarizes the articles."}]
)

response.choices[0].message.content
```

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Hi, I am Jay. How are you?"},
    {"role": "assistant", "content": "Hello Jay! I'm doing well, thank you for asking. I'm ready to assist you!\n\nPlease go ahead and provide the articles you would like me to summarize, and I will give you a concise overview of the main points."},
    {"role": "user", "content": "What's my name?"}
    ]
```

```python
response = ollama.chat.completions.create(
    model="gemma4:latest",
    messages=messages
)

response.choices[0].message.content
```
